```python
#the timeit command

import timeit
```

```python
#the output is the time it took to run this code which is an average of 14.6 microseconds
```

```python
def latitudes(size):
    latitudes = []
    x = 0
    while x < (size):
        random_lat = random.randint(-90, 90) + random.random()
        latitudes.append(random_lat)
        x += 1
    return latitudes
# Call the function with 1500.
%timeit latitudes(1500)
```
