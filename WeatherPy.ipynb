# Create Latitude and Longitude Combinations

```python
#import dependencies

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
#create a set of random latitude and longitude combinations

#to generate 1500 latitude and longitude random numbers

lats = np.random.uniform( low=-90.000, high=90.000, size=1500)
lngs = np.random.uniform(low=-180.000, high=180.000, size=1500)

#to pack latitudes and longitudes as pairs by zipping them

lat_lngs = zip(lats,lngs)
lat_lngs
```

```python
#The zip object packs each pair of lats and lngs having the same 
#index in their respective array into a tuple.
```
